```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,1,100)

def cos(x):
    return np.cos(x)
def sin(x):
    return np.sin(x)
```

```python
fig, axes = plt.subplots(1,2)

axes[0].plot(x,sin(x))
axes[0].set_title(r'$\sin(x)$')
axes[1].set_title(r'$\cos(x)$')
axes[1].plot(x,cos(x))

plt.savefig("trig.pdf")
plt.show()
```
