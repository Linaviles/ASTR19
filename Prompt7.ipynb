```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,1,100)
```

```python
def exp(x):
    return np.exp(x)
```

```python
y = exp(x)
plt.plot(x,y, color = "pink")
plt.ylabel("Awsomeness")
plt.xlabel("Time [Milliseconds]")
plt.savefig("amazing_graph.pdf")
```
